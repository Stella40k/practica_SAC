```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Defino la funcion matematica de relu
#F(x) = max(0, x)
def relu(x):
    return np.maximum(0, x)
```

```python
#Pruebo la funcion con casos especificos
calculo_negativo = -15
calculo_positivo = 10

print("--- PRUEBA DE MECANICA RELU ---")
print("Si la neurona calcula {calculo_negativo}, ReLU dice: {relu(calculo_negativo)} (Se apaga)")
print("Si la neurona calcula {calculo_positivo}, ReLU dice: {relu(calculo_positivo)} (PAsa la señal)")
```

```python
#Grafico de comportamiento general con 100 numeros q van desde -6 hasta 6
valor_x = np.linspace(-6, 6, 100)
valor_y = relu(valor_x)

plt.Figure(figsize=(7, 4))
plt.plot(valor_x, valor_y, label='f(x) = max(0, x)', color='blue', linewidth=2)
plt.title('Funcion de activacion: ReLU')
plt.xlabel('Z (calculo interno de la neurona)')
plt.ylabel('Salida de la neurona')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

Caracteristicas de ReLU:

Si el valor que recibe es negativo permanece apagado (lina en 0 en los numero negativos). Pero si es positivo los deja pasar con su mismo valor (por eso la limea sube)

Por ser matematicamente tan simple y rapido se utiliza de forma estandar en capas ocultas de redes neuronales profundas
