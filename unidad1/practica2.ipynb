```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#puntos de datos para la puerta (x1, x2)
#solo el punto (1, 1) da positivo (1)
x1 = [0, 0, 1, 1]
x2 = [0, 1, 0, 1]

colores = ["red", "red", "red", "green"] #verde = si(1),  rojo = no(0)

plt.Figure(figsize=(6, 6))
plt.scatter(x1, x2, c=colores, s=150, zorder= 3)

#dibujar la linea recta de decision generada por pesos y sesgos
#formula del material: 4*x1 + 5*x2 - 6 = 0
x_linea = np.linspace(-0.5, 1.5, 100)
y_linea = (-4 * x_linea + 6) / 5 #despejando x2 para graficar

plt.plot(x_linea, y_linea, 'b--', label="Linea de decision (Peso+Sesgo)")

#configurar el fondo del plano cartesiano
plt.axhline(0, color='black', linewidth=1) #eje X
plt.axvline(0, color='black', linewidth=1) #eje Y
plt.fill_between(x_linea, y_linea, 1.5, color='green', alpha=0.1) #zona de aprobacion
plt.fill_between(x_linea, y_linea, -0.5, color='red', alpha=0.1) #zona de rechazo

plt.xlim(-0.2, 1.2)
plt.ylim(-0.2, 1.2)
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlabel('X1 (Ej: Semaforo en verdee)')
plt.ylabel('X2 (Ej: Sin autos)')
plt.title('Separabilidad lineal en el perceptron')
plt.legend()
plt.show()
```

Perceptron Simple y Separabilidad Lineal :

Anatomia: Un PS toma multiples datos de entrada (X1, X2) los multiplica por un peso (que define la importancia) y les suma el sesgo (valor base que desplaza la frontera de decision)

Linea de Decision: la combinacion de los pesos y sesgos crea una unica linea recta capaz de cortar un plano en dos mitades exactas (Zona de activacion vs Zona de rechazo)

Separabilidad Lineal: es la capacidad de separar categorias de datos trazando una sola linea. El PS resuelve perfectamente problemas lineales (como la compuerta logica AND donde ambas condiciones deben cumplirse)

Limite Evolutivo (XOR): si los datos estan mezclados en el espacio de tal forma que es geometricamente imposible dividirlos con una unica recta, el modelo COLAPSA. Para solucionar problemas no lineales complejos, la IA evoluciono hacia MLP sumando capas ocultas para trazar multiples fronteras simultaneas
